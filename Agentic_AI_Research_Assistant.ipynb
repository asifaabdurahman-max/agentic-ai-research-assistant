```python
def create_research_plan(question):
    plan = [
        "Understand the research question",
        "Identify the main concepts",
        "Gather relevant information",
        "Check information for accuracy",
        "Prepare a clear summary"
    ]

    print("\nResearch Question:", question)
    print("\nResearch Plan:")

    for number, task in enumerate(plan, start=1):
        print(number, task)

    return plan


research_question = input("Enter your research question: ")
research_plan = create_research_plan(research_question)
```

```python
def identify_concepts(question):
    stop_words = ["what", "is", "the", "a", "an", "of", "how", "and"]

    words = question.lower().replace("?", "").split()

    concepts = []

    for word in words:
        if word not in stop_words:
            concepts.append(word)

    return concepts


key_concepts = identify_concepts(research_question)

print("Research Question:", research_question)
print("Key Concepts:", key_concepts)
```

```python
def create_summary(question, concepts):
    summary = (
        f"Your research topic is: {question}\n"
        f"The main concepts are: {', '.join(concepts)}.\n"
        "Next, gather reliable information about these concepts "
        "and verify the facts before preparing the final report."
    )

    return summary


research_summary = create_summary(
    research_question,
    key_concepts
)

print("RESEARCH SUMMARY")
print(research_summary)
```

```python
import requests
from IPython.display import display, HTML

def search_research_topic(question):
    url = "https://en.wikipedia.org/w/api.php"

    params = {
        "action": "query",
        "list": "search",
        "srsearch": question,
        "format": "json"
    }

    try:
        response = requests.get(
            url,
            params=params,
            headers={"User-Agent": "ResearchAssistant/1.0"},
            timeout=15
        )
        response.raise_for_status()

        data = response.json()
        results = data.get("query", {}).get("search", [])

        if not results:
            print("No results found. Try another question.")
            return []

        for i, result in enumerate(results[:5], start=1):
            print(f"\n{i}. {result['title']}")
            display(HTML(result.get("snippet", "")))

            print(
                "Source: https://en.wikipedia.org/wiki/"
                + result["title"].replace(" ", "_")
            )

        return results[:5]

    except requests.exceptions.Timeout:
        print("The search timed out. Please try again.")
    except requests.exceptions.RequestException as e:
        print("Network or request error:", e)
    except (ValueError, KeyError) as e:
        print("Could not process the search results:", e)

    return []


search_results = search_research_topic(research_question)
```

```python
def create_research_report(question, concepts, results):
    print("=" * 50)
    print("AGENTIC AI RESEARCH ASSISTANT")
    print("=" * 50)

    print("\n1. RESEARCH QUESTION")
    print(question)

    print("\n2. KEY CONCEPTS")
    for concept in concepts:
        print("-", concept)

    print("\n3. RESEARCH FINDINGS")
    if results:
        for i, result in enumerate(results, start=1):
            title = result.get("title", "Untitled")
            link = "https://en.wikipedia.org/wiki/" + title.replace(" ", "_")
            print(f"{i}. {title}")
            print("   Source:", link)
    else:
        print("No search results available.")

    print("\n4. NEXT ACTION")
    print("Review the sources and verify important facts.")

create_research_report(
    research_question,
    key_concepts,
    search_results
)
```

```python
def evaluate_research_results(results):
    print("RESEARCH QUALITY CHECK")

    if not results:
        print("No sources found. Search again.")
        return False

    print("Number of results:", len(results))

    for result in results:
        title = result.get("title", "Untitled")
        snippet = result.get("snippet", "")

        print("\nSource:", title)

        if snippet:
            print("Information available: Yes")
        else:
            print("Information available: Limited")

    print("\nReminder: Search results are not proof of accuracy.")
    print("Verify important claims using reliable sources.")

    return True


research_quality = evaluate_research_results(search_results)
```

```python
def prepare_ai_input(question, concepts, results):
    research_data = {
        "question": question,
        "key_concepts": concepts,
        "sources": []
    }

    for result in results:
        research_data["sources"].append({
            "title": result.get("title", "Untitled"),
            "snippet": result.get("snippet", "")
        })

    return research_data


ai_input = prepare_ai_input(
    research_question,
    key_concepts,
    search_results
)

print("AI INPUT PREPARED")
print(ai_input)
```

```python

import requests

def gather_information(topic):
    url = "https://en.wikipedia.org/w/api.php"

    params = {
        "action": "query",
        "list": "search",
        "srsearch": topic,
        "srlimit": 3,
        "format": "json"
    }

    try:
        response = requests.get(
            url,
            params=params,
            headers={
                "User-Agent": "AgenticAIResearchAssistant/1.0"
            },
            timeout=20
        )

        print("HTTP status:", response.status_code)
        response.raise_for_status()

        data = response.json()
        results = data.get("query", {}).get("search", [])

        if not results:
            print("No results found.")
            return []

        for item in results:
            print("\nTITLE:", item["title"])
            print(
                "SOURCE: https://en.wikipedia.org/?curid="
                + str(item["pageid"])
            )
            print("SUMMARY:", item["snippet"].replace(
                "<span class=\"searchmatch\">", ""
            ).replace("</span>", ""))

        return results

    except requests.RequestException as e:
        print("HTTP REQUEST ERROR:", type(e).__name__)
        print("DETAILS:", str(e))
        return []

sources = gather_information("Agentic artificial intelligence")
```

```python

def check_report_quality(project_result):
    if not project_result or not project_result.get("report"):
        print("Quality check failed: No report available.")
        return None

    report = project_result["report"]
    sources = project_result.get("sources", [])

    checks = {
        "Research question recorded": bool(project_result.get("question")),
        "Report contains an introduction": "introduction" in report.lower(),
        "Report contains findings": any(
            word in report.lower()
            for word in ["findings", "key points", "results"]
        ),
        "Report contains a conclusion": "conclusion" in report.lower(),
        "Source links available": bool(sources) and all(
            source.get("pageid") for source in sources
        )
    }

    print("RESEARCH REPORT QUALITY CHECK")
    print("=" * 35)

    for check, passed in checks.items():
        print(f"{'PASS' if passed else 'REVIEW'}: {check}")

    passed_count = sum(checks.values())
    print(f"\nChecks passed: {passed_count}/{len(checks)}")

    return checks


quality_results = check_report_quality(project_result)
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output

question_box = widgets.Textarea(
    value="",
    placeholder="Enter your research question here...",
    description="Question:",
    layout=widgets.Layout(width="100%", height="90px")
)

research_button = widgets.Button(
    description="Start Research",
    button_style="primary",
    icon="search"
)

output_box = widgets.Output()

def on_research_click(button):
    with output_box:
        clear_output(wait=True)

        question = question_box.value.strip()

        if not question:
            print("Please enter a research question.")
            return

        try:
            result = run_research_assistant(question)

            if result:
                global project_result, quality_results
                project_result = result
                quality_results = check_report_quality(result)
        except Exception as error:
            print("Something went wrong:", str(error))

research_button.on_click(on_research_click)

display(question_box, research_button, output_box)
```
